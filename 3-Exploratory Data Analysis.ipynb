```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("mobile_phones_cleaned.csv")
df.head()
```

```python
# 1) - Brand-wise Average Price

brand_price = (
    df.groupby("Brand")["Price"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=brand_price.index,
    y=brand_price.values
)

plt.xticks(rotation=45)

plt.title("Average Smartphone Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Price (₹)")
plt.tight_layout()
plt.show()
```

# Brand-wise Average Price Insight:
Apple commands the highest average smartphone price in the dataset, reflecting its premium market positioning. Google follows as the leading premium Android brand. Samsung, Nothing, OnePlus, iQOO, OPPO, and Vivo occupy the upper mid-range segment, while POCO, Redmi, Lava, and Lenovo primarily compete in the budget category. The smartphone market exhibits clear pricing segmentation, indicating diverse consumer preferences and purchasing power across product categories.

```python
# 2) - RAM Distribution
plt.figure(figsize=(8,5))

sns.countplot(
    x="RAM (GB)",
    data=df
)

plt.title("RAM Distribution")
plt.xlabel("RAM (GB)")
plt.ylabel("Number of Phones")
plt.show()
```

# RAM Distribution Insight:
The RAM distribution is centered around 8 GB, making it the most prevalent smartphone configuration in the dataset. This indicates that manufacturers consider 8 GB RAM the optimal balance between performance and affordability. While 4 GB and 6 GB variants continue to serve budget and mid-range markets, 12 GB and 16 GB RAM devices target premium users seeking enhanced multitasking and gaming performance.

```python
# 3) - Storage Distribution

plt.figure(figsize=(8,5))

sns.countplot(
    x="Storage (GB)",
    data=df
)

plt.title("Storage Distribution")
plt.xlabel("Storage (GB)")
plt.ylabel("Number of Phones")
plt.show()
```

# Storage Distribution Insight: 
The storage distribution is heavily concentrated around 128 GB, making it the most common storage configuration in the smartphone market. This indicates that manufacturers consider 128 GB the optimal balance between user requirements and device affordability. While 256 GB storage is increasingly popular among mid-range and premium smartphones, 512 GB and 1 TB variants remain niche offerings targeted at power users, content creators, and flagship device customers. The very limited presence of 32 GB and 64 GB models suggests a market-wide shift toward higher storage capacities driven by larger apps, high-resolution media, and increasing digital content consumption.

```python
# 4) - Price Segment Analysis

plt.figure(figsize=(8,5))

sns.countplot(
    x="Price_Segment",
    data=df,
    order=[
        "Budget",
        "Midrange",
        "Upper Midrange",
        "Premium"
    ]
)

plt.title("Phone Distribution by Price Segment")
plt.xlabel("Segment")
plt.ylabel("Number of Phones")
plt.show()
```

# Price Segment Analysis Insight: 
The smartphone market is primarily concentrated in the Midrange segment, which contains the largest number of devices in the dataset. Budget smartphones also represent a significant share, highlighting strong demand for affordable devices. In contrast, Upper Midrange and Premium smartphones account for a smaller proportion of the market, indicating that manufacturers focus heavily on delivering value-oriented products to attract a broader consumer base. This distribution suggests that the midrange category remains the most competitive and commercially important segment in the smartphone industry.

```python
# 5) - Camera Comparison

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Rear_Main_MP",
    y="Price",
    hue="Brand"
)

plt.title("Rear Camera MP vs Price")
plt.xlabel("Rear Camera (MP)")
plt.ylabel("Price")
plt.show()
```

# Camera Comparison Insight: 
The scatter plot indicates a general positive relationship between rear camera megapixels and smartphone price, with higher camera resolutions typically found in more expensive devices. However, the relationship is not strictly linear, as smartphones with similar camera megapixel counts are available across a wide range of prices. This suggests that pricing is influenced not only by camera specifications but also by factors such as brand value, processor performance, display quality, software features, and ecosystem offerings. Premium brands command significantly higher prices even when offering camera resolutions comparable to mid-range competitors, highlighting the importance of overall product positioning in the smartphone market.

```python
# 6) - Value Score Ranking

top_value = (
    df[
    ["Product Name","Brand","Value_Score","Price"]
    ]
    .sort_values(
        by="Value_Score",
        ascending=False
    )
    .head(15)
)

plt.figure(figsize=(12,8))

sns.barplot(
    data=top_value,
    x="Value_Score",
    y="Product Name"
)

plt.title("Top 15 Phones by Value Score")

plt.show()
print(top_value)
```

# Value Score Ranking Insight: 
The Value Score ranking highlights smartphones that offer the strongest balance between performance, customer satisfaction, and price. Apple iPhone models dominate the top positions, indicating that despite their premium pricing, they achieve exceptionally high ratings and user engagement. Budget-friendly devices such as the vivo T2x 5G and Motorola G85 5G also appear among the top-ranked phones, demonstrating that strong value can be achieved through a combination of affordability and positive customer feedback. The results suggest that perceived value is influenced not only by low price but also by brand reputation, product quality, and customer satisfaction.

```python
# 7) - Price vs RAM

plt.figure(figsize=(8,5))

sns.boxplot(
    x="RAM (GB)",
    y="Price",
    data=df
)

plt.title("Price Distribution Across RAM Sizes")
plt.show()
```

#  Price vs RAM Insight: 
The box plot shows a strong positive relationship between smartphone RAM size and price, with devices offering higher RAM capacities generally priced at higher levels. Smartphones with 3GB and 4GB RAM are concentrated in the lower price range, while 12GB and 16GB RAM devices are predominantly positioned in the premium segment. However, the increasing spread of prices within higher RAM categories indicates that RAM alone does not determine pricing. Several high-RAM smartphones are available at significantly different price points, suggesting that factors such as brand reputation, processor performance, camera quality, display technology, battery capacity, and software ecosystem also play a crucial role in pricing decisions. Additionally, the presence of multiple outliers in the 8GB, 12GB, and 16GB RAM categories highlights the existence of premium flagship devices that command substantially higher prices than typical smartphones with similar RAM specifications, emphasizing the influence of overall product positioning in the market.

```python
# 8) - Rating vs Price

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Price', y='Rating')
plt.title('Price vs Rating')

plt.show()
```

# Rating vs Price Insight:
The scatter plot suggests a moderate positive relationship between smartphone price and customer rating, with higher-priced smartphones generally receiving better ratings. Most devices, regardless of price, are rated between 4.0 and 4.7, indicating overall positive customer satisfaction across the market. However, the relationship is not perfectly linear, as smartphones within similar price ranges often have different ratings, and some expensive devices receive ratings comparable to much cheaper alternatives. This indicates that customer ratings are influenced not only by price but also by factors such as user experience, performance, camera quality, battery life, software optimization, after-sales service, and perceived value for money. Additionally, a few low-rated outliers are visible despite moderate pricing, suggesting that higher prices do not always guarantee better customer satisfaction. Overall, while premium smartphones tend to achieve slightly higher ratings, customer perception depends on the overall quality and value delivered rather than price alone.

```python
# 9) Brand-wise Average Rating (Top Brands)

top_brands = df['Brand'].value_counts().head(10).index

brand_rating = (
    df[df['Brand'].isin(top_brands)]
    .groupby('Brand')['Rating']
    .mean()
    .sort_values(ascending=False)
)

brand_rating.plot(kind='bar', figsize=(8,5))
plt.title('Average Rating by Brand')
plt.ylabel('Rating')
plt.show()
```

# Brand-wise Average Rating (Top Brands) Insight: 
The bar chart shows that average smartphone ratings are consistently high across all major brands, ranging from approximately 4.1 to 4.6. Apple holds the highest average rating, indicating strong customer satisfaction and positive user experience. Brands such as OPPO, Vivo, OnePlus, and Samsung also maintain high ratings, reflecting competitive product quality and customer acceptance. While brands like Redmi, Realme, and POCO have slightly lower average ratings, the differences are relatively small, suggesting that most brands successfully meet customer expectations. The narrow variation in ratings across brands indicates that factors such as product reliability, performance, camera quality, software experience, and value for money contribute to customer satisfaction, resulting in generally positive perceptions throughout the smartphone market. Overall, brand reputation influences ratings, but the consistently high scores suggest a highly competitive industry where most manufacturers deliver satisfactory user experiences.

```python
# Question 2: Which brand offers the highest RAM per rupee?

df['RAM_per_Rupee'] = df['RAM (GB)'] / df['Price']

df.groupby('Brand')['RAM_per_Rupee'].mean().sort_values(ascending=False)
```

# Insight:
Budget-focused brands such as POCO, Infinix, CMF, Motorola, and realme provide the highest RAM value relative to price, making them strong choices for cost-conscious consumers.
POCO and Infinix stand out as value-oriented brands that consistently offer higher memory configurations at lower price points.
    
Premium brands such as Google, Nothing, and Apple rank lower because customers pay for factors beyond RAM, including ecosystem, software experience, camera quality, and brand value.
The results indicate that RAM capacity alone is not the primary pricing driver in premium smartphone segments.

Key Takeaway: 
Brands like POCO, Infinix, and Motorola deliver the best memory value for money, while premium brands command higher prices due to overall product experience rather than RAM specifications.
